```python
import pandas as pd
```

```python
# Загрузка и проверка
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")
```

```python
print("🔍 Информация о train_df:")
print(train_df.info())
```

```python
print("\n🔍 Информация о test_df:")
print(test_df.info())
```

```python
print("📊 Пропуски в train_df:")
print(train_df.isnull().sum())
```

```python
print("\n📊 Пропуски в test_df:")
print(test_df.isnull().sum())
```

```python
# Проверяем, есть ли столбцы типа 'object'
if train_df.select_dtypes(include=['object']).shape[1] == 0:
    print("В train_df нет столбцов типа 'object'.")
else:
    print("🔍 Объектные колонки в train_df:")
    # Анализ пропущенных значений в столбцах типа 'object'
    for column in train_df.select_dtypes(include=['object']).columns:
        nan_values = train_df[column].isnull().sum()  # Подсчет числа пропущенных значений (NaN)
        unique_values = train_df[column].unique()     # Получение уникальных значений в столбце

        # Ограничение вывода уникальных значений до первых 10
        unique_values = unique_values[:10]

        print(f"В столбце '{column}' есть {nan_values} пропущенных значений (NaN)")
        print(f"Уникальные значения в столбце '{column}': {unique_values}\n")
```

```python
# Повторяем для test_df
if test_df.select_dtypes(include=['object']).shape[1] == 0:
    print("В test_df нет столбцов типа 'object'.")
else:
    print("🔍 Объектные колонки в test_df:")
    # Анализ пропущенных значений в столбцах типа 'object'
    for column in test_df.select_dtypes(include=['object']).columns:
        nan_values = test_df[column].isnull().sum()  # Подсчет числа пропущенных значений (NaN)
        unique_values = test_df[column].unique()     # Получение уникальных значений в столбце

        # Ограничение вывода уникальных значений до первых 10
        unique_values = unique_values[:10]

        print(f"В столбце '{column}' есть {nan_values} пропущенных значений (NaN)")
        print(f"Уникальные значения в столбце '{column}': {unique_values}\n")
```
